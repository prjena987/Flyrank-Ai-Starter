## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*
* **Search Question:** Which search pages are experiencing significant traffic decay and yield the highest return on editorial refresh effort?
* **Unit of Analysis:** A single published content page (URL level) evaluated over a 90-day performance window.
* **The Decision:** Which top 50 pages out of thousands should receive editorial refresh budget and rewrite bandwidth this month?
* **The Action:** Reallocate content team hours to update, expand, or re-optimize the top-ranked decaying pages.
* **Cost of a Wrong Call:**
  * **False Positive (Flagging a healthy page):** Editors waste 4-8 hours rewriting content that was already performing well, risking ranking resets and burning editorial budget.
  * **False Negative (Missing a decaying page):** Organic impressions and clicks silently collapse, resulting in compounding traffic and revenue loss before decay is manually spotted.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess
import pandas as pd
import numpy as np
REPO_URL = "https://github.com/prjena987/Flyrank-Ai-Starter.git"
TARGET_DIR = "/content/Flyrank-Ai-Starter"

if "google.colab" in sys.modules:
    if not os.path.exists(f"{TARGET_DIR}/data/raw/content_refresh_anonymized.csv"):
        print("Cloning repository into Colab environment...")
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, TARGET_DIR], check=True)
    os.chdir(TARGET_DIR)

print("Current Working Directory:", os.getcwd())

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining"] = df["trend_direction"].str.lower().eq("down").astype(int)

total_pages = len(df)
overall_declining_rate = df["is_declining"].mean() * 100

stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
stale_declining_rate = stale_visible["is_declining"].mean() * 100

print("\n--- ML-02 Metrics ---")
print(f"Metric 1 — Total Pages Analyzed: {total_pages:,}")
print(f"Metric 2 — Overall Baseline Declining Rate: {overall_declining_rate:.1f}%")
print(f"Metric 3 — Declining Rate for Stale + High-Impression Pages (>=180 days, >=500 imp): {stale_declining_rate:.1f}%")
```
